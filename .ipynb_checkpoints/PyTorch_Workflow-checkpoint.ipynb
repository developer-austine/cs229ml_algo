## 1. Data Preparation and Loading

Data can be almost anything... in machine learning.

Machine Learnig is a game of two parts:
1. Get Data into a numerical representation
2. Build a model to learn patterns in that neumrical representation.

To showcase this, let's create some **known** data using the linear reression formula

We'll use a linear regresoin formula to make a sraight line with known **parameters**

```python
# crate *known* parameters

weights  = 0.7
bias = 0.3

# create
start = 0
end = 1
step = 0.02
X = torch.arange(start, end, step).unsqueeze(dim=1)
y = weights * X + bias

X[:10], y[:10]
```

```python
len(X), len(y)
```

### Splitting the data into training and test sets (One of most important concepts in machin elearning in general

Let's Create a ta=raining and test set in our data

```python
# create a train set split
train_split = int(0.8 * len(X))
X_train, y_train = X[:train_split], y[:train_split]
X_test, y_test = X[train_split:], y[train_split:]

len(X_train), len(y_train), len(X_test), len(y_test)
```

```python
## Data visualization logic function

def plot_predictions(train_data=X_train,
                     train_labels=y_train,
                     test_data=X_test,
                     test_labels=y_test,
                     predictions=None):

    """
    Plots training data, test data and compares predictions.
    """

    plt.figure(figsize=(10, 7))

    # Plot training data in blue
    plt.scatter(train_data, train_labels, c="b", s=4, label="Training data")

    # Plot testing data in green
    plt.scatter(test_data, test_labels, c="g", s=4, label="Testing data")

    # Are there predictions?
    if predictions is not None:
        # plot the pedictions if they exist
        plt.scatter(test_data, predictions, c="r", s=4, label="predictions")

    # show the legend
    plt.legend(prop={"size": 14});
```

```python
plot_predictions()
```

## 2. Build  Model

Our first PyTorch Model!

this is very exciting.....

What our model does:
* start with random values (weights and bias)
* Look at the raining data and adjust random values to better represent (or get closer to)
  the ideal values (the weight & bias values we used to create the data)

How does it do so?

Through two main algorithms:
1. Gradient Descent
2. BackPropagation.

```python
from torch import nn

# create a linear regression model class
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.weights = nn.Parameter(torch.randn(1, # <- start with a random weight and try to adjust it to the ideal weight
                                               requires_grad=True, # <- Can this parameter be updated via gradient Descent?
                                               dtype=torch.float)) # <- PyTorch loves the datatype float32

        self.bias = nn.Parameter(torch.randn(1, # <- start with a random bias and try to adjust it to the ideal bias
                                            requires_grad=True, # <- Can this parameter be updated via gradient Descent?
                                            dtype=torch.float)) # <- PyTorch loves the datatype float32

    # Forward method defines the computation of the model
    def forward(self, x: torch.Tensor) -> torch.Tensor: # X is the input data
        return self.weights * x + self.bias # This is the linear regression formula
```

 ## 3. Training our very frst machine learning model

 Our whole idea of training is for a model to move to some unknown parametrs (these may be rnadom) to some known parameters, or in ther words, from a poor rpresentation of data to a batter representation of it.

 One way to measure how poor ot howwrong the model predicitons are is to use a `loss function`

 * Note: Loss function may also be called cost function or criterion in dfferent areas

   ## What to Train

 * **Loss function** is a function to measure how wrong your pedictions are to the idela data.

 * **Optimizer** takes into account the loss of a model and adjusts the model's parameters (e.g weights and bias) to improve the loss function

And specifically in pytorch, we need:
*  A training loop
*  A testing loop

```python
# check out our model parameter, a parameter is a value a model sets up itself
model_0.state_dict()
```

### Setting up loss function and the Optimizer

```python
# Setup a loss function
# Manual loss calculation
# MAE_Loss = torch.mean(torch.abs(y_pred-y_test))
loss_fn = nn.L1Loss() # PyTorch Workflow for the L1 Loss

#setup and Optimizer SGD (Stochastic Gradient Descent)
optimizer = torch.optim.SGD(model_0.parameters(),
                           lr=0.01) # lr - Learning rate - possibly the most important hyperparameter
```

```python
loss_fn
```

### Building a Training loop (and a testing loop) in PyTorch

A couple of things we need in a trainig loop:
0.  Loop through the data
1.  Forward Pass (this involves data moving through our model's `forward()` functions) to make peditions on data - also called forward propagation
2.  Calculate the loss (compare forward pass predictions to ground truth levels)
3.   Optimizer zero grad
4.   Loss Backwards - move backwards hrough the network to calculate the gradient of each of the parameters of our model with respect to the loss (**backpropagation**)
5.   Optimzer step - use the optimizer to adjust our model's parameters to try and improve the loss (**gradient descent**)

```python
torch.manual_seed(42)

# An epoch is one loop through the data...
epochs = 200

epoch_count = []
loss_values = []
test_loss_values = []

### Training
# step 0 - Loop through the data...
for epoch in range(epochs):
    # Set the model to training mode
    model_0.train() # tarin mode in PyTorch set all parameters that reuire gradients to require gradients

    # 1. Forward Pass
    y_pred = model_0(X_train)

    # 2. Calculate the loss
    loss = loss_fn(y_pred, y_train)
    #print(f"Loss: {loss}")

    # 3. Optimizer zero grad
    optimizer.zero_grad()

    # 4. Perform backpropagation on the loss wrt parameters of the model
    loss.backward()

    # 5. Step the optimizer (perform gradient descent)
    optimizer.step() # by default how the optimizer changes will accumulate through the loop so...we have to zero them above in step 3 for the next iteration of the loop. 

    ### Testing
    model_0.eval() # turns off different settings in the model not needed for evaluation or testing
    with torch.inference_mode(): #turns off gradient tracking and a couple more things behind the scene
        # 1. Do the forward pass
        test_pred = model_0(X_test)

        # 2. Calculate the loss
        test_loss = loss_fn(test_pred, y_test)

    # Print out what's happening
    if epoch % 10 == 0:
        epoch_count.append(epoch)
        loss_values.append(loss)
        test_loss_values.append(test_loss)
        print(f"Epoch: {epoch} | Loss: {loss} | Test Loss: {test_loss} ")
        
    # Print out model state_dict()
    print(model_0.state_dict())
```

```python
with torch.inference_mode():
    y_preds_new = model_0(X_test)
```

```python
model_0.state_dict()
```

```python
import numpy as np
np.array(torch.tensor(loss_values).numpy()), test_loss_values
```

```python
# plot the loss curves
plt.plot(epoch_count, np.array(torch.tensor(loss_values).numpy()), label="Train Loss")
plt.plot(epoch_count, test_loss_values, label="Test Loss")
plt.title("Training and Test loss curves")
plt.ylabel("Loss")
plt.xlabel("Epochs")
plt.legend()
```

```python
plot_predictions(predictions=y_preds_new)
```

## Saving a Model in PyTorch

There ae three main methods ofr saving and loading models in pytorch.

1. `torch.save()` - allows you to save object in python's pickle format
2. `torch.load()` - allows you to load a saved pytorch object
3. `torch.nn.Module.load_state_dict()` - allows to load a model's saved state dict

```python
# saving a pytorch model
from pathlib import Path

# 1. create a model directory
MODEL_PATH = Path("models")
MODEL_PATH.mkdir(parents=True, exist_ok=True)

# 2. Create a models save path
MODEL_NAME = "01_pytorch_workflow_model_0.pth"
MODEL_SAVE_PATH = MODEL_PATH / MODEL_NAME

# 3. Save the model state dict
print(f"Saving model to: {MODEL_SAVE_PATH}")
torch.save(obj=model_0.state_dict(),
           f=MODEL_SAVE_PATH)
```
