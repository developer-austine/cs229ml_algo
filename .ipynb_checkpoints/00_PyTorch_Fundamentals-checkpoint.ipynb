### Matrix Multliplication

Two main ways of performing matrix multliplication in neural networks and deep learning

1. Element-Wise multiplication
2. Matrix Multiplication


there are two main rules that performing matrix multiplication needs to satisfy
1. The **inner dimension** must match:
* `(3, 2) @ (3, 2)` won't work
* `(2, 3) @ (3, 2)` will work
* `(3, 2) @ (2, 3)` will work
  
2. The resulting matrix has the shape of the **outer dimensions**:
* `(2, 3) @ (3, 2)` -> `(2, 2)`
* `(3, 2) @ (2, 3)` -> `(3, 3)` 

```python
torch.matmul(torch.rand(3, 2), torch.rand(2, 3))
```

### one of the most common errors in deep learning

```python
# shapes for matrix multiplication 
tensor_A = torch.tensor([[1,2],
                         [3, 4],
                         [5, 6]])

tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])

code_answer = torch.matmul(tensor_A, tensor_B) #torch.mm is the same as torch.matmul (it is an alias for writing less code)
code_answer
```

```python
tensor_A.shape, tensor_B.shape
```

To fix our tesnor shapemissues, we can manipulate the shap of one of our tensors using a **transpose**

A **transpose** switches the axes or dimensions of a gven tensor.

```python
new_tensor_B = tensor_B.T
new_tensor_B
```

```python
tensor_B
```

```python
new_tensor_B.shape
```

```python
tensor_A @ new_tensor_B # This will now work as the matrix B has been transposed.
```

```python
# The matrix multiplicaton woperation works when tensor_B is transposed
print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")
print(f"New shapes: tensor_A = {tensor_A.shape} (same shape as above), tensor_B.T = {new_tensor_B.shape}")
print(f"New shape after maultiplication:{(tensor_A @ new_tensor_B).shape}")
```

## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshapes an input tensor to a definite shape
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - combine multiple tensors on top of each other (vstack) or side by side (hstack)
* Squeeze - remove all `1` dimensions from a tensor
* unsqueeze - add a `1` dimension to the tensor
* permite - Return a view of the input with dimensions permutted (swapped) in a certain way

```python
# let us create a tensor 

import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Add an extra dimension
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
# change the view
z = x.view(1, 9)
z, z.shape
```

```python
# changing z changes x (because a view of a tensor shares the same memory as the original input)
z[:, 0] = 5
z, x
```

```python
# stack tensors on top of each other (1 stackes them vertically while 0 stacks them horizontally)
x_stacked = torch.stack([x, x, x, x], dim=0)
x_stacked
```

```python
# torch.squeeze removes all single dimensions from a target tensor
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

# Remove extra dimensions from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_reshaped}")
print(f"New shape: {x_squeezed.shape}")
```

```python
# torch.unsqueeze()- adds a single dimension to a target at a specific dim(dimension)
print(f"Previous target: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

#Add a  extra dimenion with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# torch.permute -rearanges the dimenions of a target tensor in a specified order

x_original = torch.rand(size=(224, 224, 3)) # [height, width, color_channels]

# permute the original tensor to rearange the axis (or dim) order
x_permuted = x_original.permute(2, 0, 1) # shifts the axis 0 -> 1, 1-> 2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shae: {x_permuted.shape}") # [color_channels, height, width]
```

```python
x_original[:, :, 0]
```

```python
x_original[0, 0, 0] = 728218
x_original[0, 0, 0], x_permuted[0, 0, 0]
```

## Pytorch Tensors and Nunpy

NumPy is a popular scientific python numerical computing library.

And because of ths, pytorch has functionality to interact with it.

* Data in NumPy, want in Pytorch tensor -> `torch.from_numpy(ndarray)`
* PyTorch tensor -> NumPy -> `torch.tensor.numpy()`

```python
# NumPy array to Tensor
import torch
import numpy as np

array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array) # Warning: When converting from numpy -> pytorch, pytorch refects numpy's default datatype of float64 unless specified otherwise
array, tensor
```

```python
# Change the value of the array
array = array + 1
array, tensor
```

```python
tensor.dtype
```

```python
# Tensor to numPy array
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

```python
# Chnag the tensor, what happens to `numpy_tensor?`
tensor = tensor + 1
tensor, numpy_tensor
```

## Pytorch Reproducibility (Taking the random out of random)

In short how a network learns:

Start with reandom numbers -> tensor oerations -> update random numners to try and make them better representation of the data -> again -> again -> again

To reduce the rndomness in neural networsk and pytorch, comes the concept of a **random seed**

Essentially what the random seed does is "flavour" the randomness.

```python
# create a random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Let's make some random but reproducibile tensors
import torch

# Set random seed
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
